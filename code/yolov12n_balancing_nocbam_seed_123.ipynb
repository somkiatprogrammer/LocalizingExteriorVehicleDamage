```python
# Initialise
from ultralytics import YOLO

# Code to ignore warnings from function usage
import warnings
warnings.filterwarnings('ignore')
```

```python
# Define image size
imgsz = 640
```

```python
# Training
model = YOLO("../yolo/yolo12n.pt")
results = model.train(
    data="dataset.yaml", 
    epochs=200,
    imgsz=imgsz, 
    iou=0.6,
    box=9.60,
    lr0=0.0055,
    cls=2.00,
    patience=50,
    mosaic=0.58,
    mixup=0.22,
    plots=True,
    augment=True,
    resume=False,
    device=0,
    seed=123
)
```

```python
%%time

# Test evaluation
model = YOLO("./runs/detect/train521/weights/best.pt")
metrics = model.val(
    data="dataset.yaml", 
    imgsz=imgsz, 
    conf=0.25, 
    iou=0.6, 
    split="test", 
    plots=True, 
    augment=False, 
    save_json=True, 
    device=0
)
```
